#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r

library(tidymodels)

# Use the same seed as Task 0.1
set.seed(40764979)

# Select the model variables and remove missing values
basketball_model_data<-basketball_tidy|>
select(conf_wins_pct,tourney_wins)|>
filter(!is.na(conf_wins_pct), !is.na(tourney_wins))

# Split the data into 75% training and 25% testing
basketball_split<-initial_split(basketball_model_data,prop=0.75)

# Extract the training and test data
basketball_train<-training(basketball_split)
basketball_test<-testing(basketball_split)

# Specify a linear regression model
basketball_spec<-linear_reg()|>
set_engine("lm")|>
set_mode("regression")

# Set the outcome and predictor
basketball_recipe<-recipe(tourney_wins~conf_wins_pct,data=basketball_train)

# Fit the model using only the training data
basketball_fit<-workflow()|>
add_recipe(basketball_recipe)|>
add_model(basketball_spec)|>
fit(data=basketball_train)

# Predict test outcomes and attach the actual values
basketball_predictions<-predict(basketball_fit,new_data=basketball_test)|>
bind_cols(basketball_test)

# Calculate the test RMSE
basketball_rmse<-basketball_predictions|>
metrics(truth=tourney_wins,estimate=.pred)|>
filter(.metric=="rmse")

# Display the test RMSE
basketball_rmse
```
